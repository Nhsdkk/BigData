### Поля sale_product_id, sale_customer_id, sale_seller_id имеют всего 1000 уникальных значений среди всех батчей и не равное 1000 количество уникальных значений в каждом батче, что указывает, что несколько продаж ссылаются на 1 продукт. Причем во всем датасете эти id от 1 до 1000, следовательно продажа ссылается на продукт внутри своего батча.

```python
product_id_unique_in_batch = True
customer_id_unique_in_batch = True
seller_id_unique_in_batch = True

for i in range(0, 10000, 1000):
    product_id_unique_in_batch = product_id_unique_in_batch and data['sale_product_id'][i:i+1000].nunique() == 1000
    customer_id_unique_in_batch = customer_id_unique_in_batch and data['sale_customer_id'][i:i+1000].nunique() == 1000
    seller_id_unique_in_batch = seller_id_unique_in_batch and data['sale_seller_id'][i:i+1000].nunique() == 1000

(product_id_unique_in_batch, customer_id_unique_in_batch, seller_id_unique_in_batch)
```

### Видно, что у каждого клиента есть 1 питомец, однако невозможно определить уникальный набор атрибутов для них, соответственно уникальным ключем для питомца будет сгенерированный id.

```python
%%sql
SELECT mock_data.customer_pet_type, mock_data.customer_pet_name, mock_data.customer_pet_breed, COUNT(*)
FROM mock_data
GROUP BY mock_data.customer_pet_type, mock_data.customer_pet_name, mock_data.customer_pet_breed
ORDER BY count(*) DESC
```
